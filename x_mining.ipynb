```python
# Installation: pip install ntscraper pandas
from ntscraper import Nitter
import pandas as pd

def scrape_hashtag_data(hashtag, count=10):
    scraper = Nitter()
    print(f"Suche nach #{hashtag}...")
    
    # Suche nach Hashtag, 'mode' kann 'hashtag' oder 'term' sein
    results = scraper.get_tweets(hashtag, mode='hashtag', number=count)
    
    processed_tweets = []
    for tweet in results['tweets']:
        processed_tweets.append({
            'date': tweet['date'],
            'text': tweet['text'],
            'user': tweet['user']['name'],
            'likes': tweet['stats']['likes'],
            'hashtags': [hashtag] # Wir wissen, wonach wir gesucht haben
        })
    return processed_tweets

# Daten für beide Hashtags sammeln
data_trump = scrape_hashtag_data("Trump", 10)
data_venezuela = scrape_hashtag_data("Venezuela", 10)

# Zusammenführen und Speichern
df = pd.DataFrame(data_trump + data_venezuela)
df.to_csv("twitter_data_analysis.csv", index=False, encoding='utf-8')
print("Datei 'twitter_data_analysis.csv' wurde erstellt!")
```
